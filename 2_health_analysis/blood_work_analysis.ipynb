```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()

llm = ChatGroq(model="qwen/qwen3-32b",temperature=0.4)
```

```python
with open("blood_work.txt", "r") as f:
    blood_report = f.read()

print(blood_report[:200])
```

```python
extraction_prompt = f"""
you are a medical data extraction assistant.ChatGroq

from the blood report below , extract all test values and classify each one as HIGH, LOW, OR NORMAL based on the reference range provided in the report.

format your response as :
- Test Name : value | Status (HIGH, LOW, NORMAL) | reference range

Blood Report:
{blood_report}
"""
```

```python
extraction_response = llm.invoke(extraction_prompt)
extracted_values = extraction_response.text
print(extracted_values)
```

```python
diet_prompt = f"""
Based on the extracted blood test results below, provide a personalized diet recommendation to help improve any abnormal values.

based on the blood work analysis below, write:
1. a short summary in 3 lines explaining the patient's current health status.
2. A short , practical indian diet plan having only two sections (1) foods to avoid (2) foods to eat more of.
Do not include any other sections in the diet plan.

Blood Test Analysis:
{extracted_values}
"""
```

```python
diet_response = llm.invoke(diet_prompt)

print(diet_response.text)
```
